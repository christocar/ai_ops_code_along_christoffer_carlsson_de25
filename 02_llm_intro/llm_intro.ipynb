# Intro to LLM and PydanticAI

```python
from pydantic_ai import Agent
from pydantic_ai.models.openai import OpenAIChatModel
from pydantic_ai.providers.openai import OpenAIProvider
import os
from dotenv import load_dotenv

load_dotenv()

model = OpenAIChatModel(
    model_name="nvidia/nemotron-3.5-lightning-30b-a3b",
    provider=OpenAIProvider(
        base_url = "https://integrate.api.nvidia.com/v1", api_key=os.environ["NVIDIA_API"]
    ),
)


model
```

## ChefBot

```python
chef_agent = Agent(
    model=model,
    instructions="You are a professional chef that is very good at making asian food and explaining in a pedagogical way how to create yummy dishes."
                  """
                    Answer in the following format:

                    Description 1-2 sentences of what the dish is
        
                    List of ingredients in bullet points

                    Short step by step description on how to make the dish
                  """ 
)
result = await chef_agent.run("Make a list of three asian soups", timeout=60)
print(result.output)
```

```python
chef_agent = Agent(
    model=model,
    instructions="You are a professional chef that is very good at making asian food and explaining in a pedagogical way how to create yummy dishes."
                  """
                    Answer in the following format:

                    Description 1-2 sentences of what the dish is
        
                    List of ingredients in bullet points

                    Short step by step description on how to make the dish
                  """ 
)
result = await chef_agent.run("""
Make a list of three Asian soups from three different countries.

Return ONLY a Python list of dictionaries, with no text before or after.
Each dictionary must have exactly these keys:
- "name": the name of the soup (string)
- "country": country of origin (string)
- "description": one sentence describing the soup (string)
- "ingredients": ingredients with amounts for 4 servings, e.g. "200 g shrimp" (list of strings)
- "steps": cooking steps, one step per item (list of strings)

Use the ingredients that are traditional for each dish.
""")
print(result.output)
```
