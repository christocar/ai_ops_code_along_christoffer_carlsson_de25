# Pydantic with Nvidia Build

```python
from pydantic_ai import Agent
from pydantic_ai.models.openai import OpenAIChatModel
from pydantic_ai.providers.openai import OpenAIProvider
import os
from dotenv import load_dotenv

load_dotenv()

model = OpenAIChatModel(
    model_name="nvidia/nemotron-3.5-lightning-30b-a3b",
    provider=OpenAIProvider(
        base_url = "https://integrate.api.nvidia.com/v1", api_key=os.environ["NVIDIA_API"]
    ),
)


model
```

```python
agent = Agent(model)
result = await agent.run("Give me a funny python joke", timeout=60)
result
```

```python
model = OpenAIChatModel(
    model_name="nvidia/nemotron-3.5-lightning-30b-a3b",
    provider=OpenAIProvider(
        base_url = "https://integrate.api.nvidia.com/v1", api_key=os.environ["NVIDIA_API"]
    ),
)


model
```

```python
agent = Agent(model = model, system_prompt="You are a funny joking bot, answer shortly in a nerdy style, also add emojis.")

prompt = "Explain what PydanticAI is and the difference between it and Pydantic"
result = await agent.run(prompt, timeout=60)
print(result.output)    
```
